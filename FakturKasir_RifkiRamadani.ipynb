```python
# ==============================================================================
# Nama : Rifki Ramadani
# NIM : 26552011042
# Matkul : Pengenalan Pemrograman
# Ketentuan : Formatting f-string alignment 36 karakter (Header ^, Label <, Nominal >)
# ==============================================================================

from datetime import datetime

def format_rupiah(nominal: float) -> str:
    """Mengubah nilai angka desimal menjadi format Rupiah dengan koma ribuan dan 2 desimal."""
    return f"Rp{nominal:,.2f}"

def main():
    LEBAR_NOTA = 36

    print("=" * LEBAR_NOTA)
    print(f"{'SISTEM KASIR':^{LEBAR_NOTA}}")
    print("=" * LEBAR_NOTA)

    nama_pelanggan = input("Masukkan Nama Pelanggan : ").strip()
    if not nama_pelanggan:
        nama_pelanggan = "Pelanggan Umum"

    while True:
        try:
            total_awal = float(input("Total Belanja Sebelum Diskon (Rp) : "))
            if total_awal < 0:
                print("Error: Nominal belanja tidak boleh negatif!")
                continue
            break
        except ValueError:
            print("Error: Harap masukkan angka yang valid!")

    while True:
        try:
            persen_diskon = float(input("Masukkan Persentase Diskon (%)    : "))
            if persen_diskon < 0 or persen_diskon > 100:
                print("Error: Diskon harus berada di rentang 0 - 100%!")
                continue
            break
        except ValueError:
            print("Error: Harap masukkan angka yang valid!")

    while True:
        try:

            nominal_diskon_temp = total_awal * (persen_diskon / 100)
            total_bayar_temp = total_awal - nominal_diskon_temp

            uang_bayar = float(input(f"Masukkan Nominal Tunai Bayar (Rp)  : "))
            if uang_bayar < total_bayar_temp:
                print(f"Error: Uang tunai kurang! Minimal bayar: {format_rupiah(total_bayar_temp)}")
                continue
            break
        except ValueError:
            print("Error: Harap masukkan angka yang valid!")

    nominal_diskon = total_awal * (persen_diskon / 100)
    total_bayar = total_awal - nominal_diskon
    kembalian = uang_bayar - total_bayar
    waktu_transaksi = datetime.now().strftime("%d/%m/%Y %H:%M")


    lbl_total_awal = "Total Awal"
    val_total_awal = format_rupiah(total_awal)


    diskon_str = f"{persen_diskon:.1f}" if persen_diskon % 1 != 0 else f"{int(persen_diskon)}"
    lbl_diskon = f"Diskon ({diskon_str}%)"
    val_diskon = format_rupiah(nominal_diskon)

    lbl_total_bayar = "Total Bayar"
    val_total_bayar = format_rupiah(total_bayar)

    lbl_tunai = "Tunai"
    val_tunai = format_rupiah(uang_bayar)

    lbl_kembali = "Kembalian"
    val_kembali = format_rupiah(kembalian)

    print("\n" + "=" * LEBAR_NOTA)


    print(f"{'Kinndara Mart':^{LEBAR_NOTA}}")
    print(f"{'NOTA PEMBAYARAN':^{LEBAR_NOTA}}")
    print("=" * LEBAR_NOTA)

    print(f"{f'Pelanggan: {nama_pelanggan}':^{LEBAR_NOTA}}")
    print(f"{f'Tgl: {waktu_transaksi}':^{LEBAR_NOTA}}")
    print("-" * LEBAR_NOTA)


    print(f"{lbl_total_awal:<16}{val_total_awal:>20}")
    print(f"{lbl_diskon:<16}{val_diskon:>20}")
    print("-" * LEBAR_NOTA)
    print(f"{lbl_total_bayar:<16}{val_total_bayar:>20}")
    print(f"{lbl_tunai:<16}{val_tunai:>20}")
    print(f"{lbl_kembali:<16}{val_kembali:>20}")

    print("=" * LEBAR_NOTA)
    print(f"{'Terima Kasih Atas Kunjungan Anda!':^{LEBAR_NOTA}}")
    print("=" * LEBAR_NOTA + "\n")

if __name__ == "__main__":
    main()
```
