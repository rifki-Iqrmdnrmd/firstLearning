```python
import math

class Balok:
    # Variabel statis (Class variables di Python)
    jumVolumeTotalBalok = 0.0
    jumBalok = 0
    Volume = 0.0
    luasPermukaan = 0.0
    diagonalRuang = 0.0
    digSisiDepan = 0.0
    digSisiSamping = 0.0
    digSisiAtas = 0.0

    def __init__(self):
        # Variabel instance
        self.panjang = 0.0
        self._lebar = 0.0       # protected (menggunakan satu underscore)
        self.tinggi = 0.0

    def hitungVolume(self):
        Balok.Volume = self.panjang * self._lebar * self.tinggi
        return Balok.Volume

    def hitungLuasPermukaan(self):
        Balok.luasPermukaan = 2 * (self.panjang * self._lebar + self.panjang * self.tinggi + self.tinggi * self._lebar)
        return Balok.luasPermukaan

    def hitungDiagonalRuang(self):
        Balok.diagonalRuang = math.sqrt((self.hitungDiagonalSisiDepan() ** 2) + (self._lebar ** 2))
        return Balok.diagonalRuang

    def hitungDiagonalSisiDepan(self):
        Balok.digSisiDepan = math.sqrt((self.panjang ** 2) + (self.tinggi ** 2))
        return Balok.digSisiDepan

    def _hitungDiagonalSisiSamping(self):
        Balok.digSisiSamping = math.sqrt((self._lebar ** 2) + (self.tinggi ** 2))
        return Balok.digSisiSamping

    def __hitungDiagonalSisiAtas(self):
        Balok.digSisiAtas = math.sqrt((self.panjang ** 2) + (self._lebar ** 2))
        return Balok.digSisiAtas

    # Karena Python menerapkan strict name mangling pada private methods (__),
    # kita buatkan helper pemanggil untuk digunakan di "main" agar tidak error.
    def panggilDiagonalSisiAtas(self):
        return self.__hitungDiagonalSisiAtas()

    @classmethod
    def jumTotalVolume(cls):
        cls.jumVolumeTotalBalok += cls.Volume
        return cls.jumVolumeTotalBalok

    @classmethod
    def tamJumlahBalok(cls):
        cls.jumBalok += 1

    def tampilData(self):
        print(f"Panjang               : {self.panjang}")
        print(f"Lebar                 : {self._lebar}")
        print(f"Tinggi                : {self.tinggi}")


# Eksekusi Main Method
if __name__ == "__main__":
    b1 = Balok()
    b1.panjang = 4
    b1._lebar = 2
    b1.tinggi = 3

    b1.hitungVolume()
    b1.hitungLuasPermukaan()
    b1.hitungDiagonalSisiDepan()
    b1.hitungDiagonalRuang()
    b1._hitungDiagonalSisiSamping()
    b1.panggilDiagonalSisiAtas()
    Balok.jumTotalVolume()
    Balok.tamJumlahBalok()

    b1.tampilData()
    print(f"Volume                : {b1.hitungVolume()}")
    print(f"Luas Permukaan        : {b1.hitungLuasPermukaan()}")
    print(f"Diagonal Ruang        : {b1.hitungDiagonalRuang()}")
    print(f"Diagonal Sisi Depan   : {b1.hitungDiagonalSisiDepan()}")
    print(f"Diagonal Sisi Samping : {b1._hitungDiagonalSisiSamping()}")
    print(f"Diagonal Sisi Atas    : {b1.panggilDiagonalSisiAtas()}")

    b2 = Balok()
    b2.panjang = 5
    b2._lebar = 3
    b2.tinggi = 4

    b2.hitungVolume()
    b2.hitungLuasPermukaan()
    b2.hitungDiagonalSisiDepan()
    b2.hitungDiagonalRuang()
    b2._hitungDiagonalSisiSamping()
    b2.panggilDiagonalSisiAtas()
    Balok.tamJumlahBalok()

    b2.tampilData()
    print(f"Volume                : {b2.hitungVolume()}")
    print(f"Luas Permukaan        : {b2.hitungLuasPermukaan()}")
    print(f"Diagonal Ruang        : {b2.hitungDiagonalRuang()}")
    print(f"Diagonal Sisi Depan   : {b2.hitungDiagonalSisiDepan()}")
    print(f"Diagonal Sisi Samping : {b2._hitungDiagonalSisiSamping()}")
    print(f"Diagonal Sisi Atas    : {b2.panggilDiagonalSisiAtas()}")
    print(f"JUMLAH SEMUA VOLUME BALOK    : {Balok.jumTotalVolume()}")
    print(f"JUMLAH BALOK YANG DIBUAT     : {Balok.jumBalok}")
```
